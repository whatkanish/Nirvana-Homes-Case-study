# NIRVANA HOMES CASE STUDY ANALYSIS

QUESTION 3: SALES-SPEED CLASSIFICATION



```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score, accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.dummy import DummyClassifier
from sklearn.preprocessing import StandardScaler


# Set random state
random_state = 42

# ------------------------------------------------------------------------
# STEP 1: LOAD AND AUDIT THE DATA
# ------------------------------------------------------------------------
file_path = '/content/nirvana_homes_listings - Sheet1.csv'
df = pd.read_csv(file_path)

print('=== Dataset Shape ===')
print(df.shape)
print('\n=== Column Names & Data Types ===')
print(df.dtypes)
print('\n=== First 5 Rows ===')
display(df.head())
print('\n=== Missing Values ===')
print(df.isnull().sum())
print('\n=== Exact Duplicate Rows ===')
num_duplicates = df.duplicated().sum()
print(f'Number of exact duplicate rows: {num_duplicates}')

# Handle duplicates if they exist
if num_duplicates > 0:
    df = df.drop_duplicates()
    print(f'Duplicates removed. New shape: {df.shape}')

print('\n=== Summary Statistics ===')
display(df.describe(include='all'))
```

```python
# Print distributions of target variables
print('=== Distribution of listed_price_lakhs ===')
print(df['listed_price_lakhs'].describe())
print('\n=== Distribution of sold_within_45_days ===')
print(df['sold_within_45_days'].value_counts(dropna=False))
```

```python
# ------------------------------------------------------------------------
# QUESTION 1: PRICING MODEL
# ------------------------------------------------------------------------

# Define allowed predictors
predictors_reg = [
    'city', 'locality_tier', 'bhk', 'area_sqft', 'age_years', 'floor_number',
    'total_floors', 'distance_to_metro_km', 'amenity_score', 'furnishing',
    'parking_spots', 'is_corner_unit', 'maintenance_monthly_inr',
    'broker_experience_years', 'num_photos_uploaded',
    'listing_agent_photo_quality_score', 'festival_season_listing'
]

X_reg_raw = df[predictors_reg].copy()
y_reg = df['listed_price_lakhs'].copy()

# Handle missing amenity_score
# Check if amenity_score has missing values, fill with median or mean
if X_reg_raw['amenity_score'].isnull().sum() > 0:
    median_amenity = X_reg_raw['amenity_score'].median()
    print(f'Imputing missing amenity_score with median: {median_amenity}')
    X_reg_raw['amenity_score'] = X_reg_raw['amenity_score'].fillna(median_amenity)

# Handle categorical columns using pd.get_dummies, keeping track of reference categories
categorical_cols = ['city', 'locality_tier', 'furnishing', 'festival_season_listing']
# To clearly identify the reference categories, let us use drop_first=True
X_reg = pd.get_dummies(X_reg_raw, columns=categorical_cols, drop_first=True, dtype=float)

# Split train and test
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg, y_reg, test_size=0.20, random_state=random_state
)

# Fit multiple linear regression
reg_model = LinearRegression()
reg_model.fit(X_train_reg, y_train_reg)

# Evaluate model
def evaluate_reg_model(model, X, y):
    preds = model.predict(X)
    r2 = r2_score(y, preds)
    rmse = np.sqrt(mean_squared_error(y, preds))
    mae = mean_absolute_error(y, preds)
    return r2, rmse, mae, preds

train_r2, train_rmse, train_mae, train_preds = evaluate_reg_model(reg_model, X_train_reg, y_train_reg)
test_r2, test_rmse, test_mae, test_preds = evaluate_reg_model(reg_model, X_test_reg, y_test_reg)

print('\n=== Model Performance Table ===')
perf_table = pd.DataFrame({
    'Metric': ['R²', 'RMSE', 'MAE'],
    'Training': [train_r2, train_rmse, train_mae],
    'Testing': [test_r2, test_rmse, test_mae]
})
display(perf_table)

# Coefficients Table
coef_df = pd.DataFrame({
    'Feature': X_reg.columns,
    'Coefficient': reg_model.coef_
})
print(f'Intercept: {reg_model.intercept_}')
print('\n=== Coefficients ===')
display(coef_df.sort_values(by='Coefficient', ascending=False))
```

```python
# To find the genuine strongest price effects (taking units into account),
# let's look at standardized coefficients or coefficients scaled by standard deviation of features.
feature_stds = X_train_reg.std()
coef_df['Standardized_Coefficient'] = coef_df.apply(
    lambda row: row['Coefficient'] * feature_stds[row['Feature']], axis=1
)
coef_df['Absolute_Standardized_Coefficient'] = coef_df['Standardized_Coefficient'].abs()
print('\n=== Features Ranked by Standardized Coefficient (Absolute Impact) ===')
display(coef_df.sort_values(by='Absolute_Standardized_Coefficient', ascending=False))
```

```python
# ------------------------------------------------------------------------
# QUESTION 2: PRICING MODEL ERROR ANALYSIS
# ------------------------------------------------------------------------

# Align indices and build dataframe for testing set prediction errors
test_indices = X_test_reg.index
test_analysis_df = df.loc[test_indices].copy()
test_analysis_df['actual_price'] = y_test_reg
test_analysis_df['predicted_price'] = test_preds
test_analysis_df['error'] = test_analysis_df['actual_price'] - test_analysis_df['predicted_price']
test_analysis_df['absolute_error'] = test_analysis_df['error'].abs()

# Top 10 largest absolute errors
top_10_errors = test_analysis_df.sort_values(by='absolute_error', ascending=False).head(10)
print('=== Top 10 Largest Absolute Errors ===')
display(top_10_errors[[
    'listing_id', 'city', 'locality_tier', 'bhk', 'area_sqft', 'age_years',
    'actual_price', 'predicted_price', 'absolute_error'
]])
```

```python
# Define Large Flats and Old Buildings based on upper quartiles
upper_quartile_area = df['area_sqft'].quantile(0.75)
upper_quartile_age = df['age_years'].quantile(0.75)

print(f"Upper quartile of area_sqft: {upper_quartile_area}")
print(f"Upper quartile of age_years: {upper_quartile_age}")

test_analysis_df['is_large_flat'] = test_analysis_df['area_sqft'] >= upper_quartile_area
test_analysis_df['is_old_building'] = test_analysis_df['age_years'] >= upper_quartile_age

# Function to calculate errors for groups
def calculate_group_errors(dataframe, group_col, group_name_prefix=''):
    grouped = dataframe.groupby(group_col)
    summary = []
    for name, group in grouped:
        n = len(group)
        mae = mean_absolute_error(group['actual_price'], group['predicted_price'])
        rmse = np.sqrt(mean_squared_error(group['actual_price'], group['predicted_price']))
        mean_err = group['error'].mean()
        summary.append({
            'Category': f"{group_name_prefix}{name}",
            'N': n,
            'MAE': mae,
            'RMSE': rmse,
            'Mean_Error': mean_err
        })
    return summary

category_summaries = []
category_summaries.extend(calculate_group_errors(test_analysis_df, 'city', 'City: '))
category_summaries.extend(calculate_group_errors(test_analysis_df, 'locality_tier', 'Locality Tier: '))
category_summaries.extend(calculate_group_errors(test_analysis_df, 'bhk', 'BHK: '))
category_summaries.extend(calculate_group_errors(test_analysis_df, 'furnishing', 'Furnishing: '))
category_summaries.extend(calculate_group_errors(test_analysis_df, 'is_corner_unit', 'Corner Unit: '))
category_summaries.extend(calculate_group_errors(test_analysis_df, 'is_large_flat', 'Large Flat: '))
category_summaries.extend(calculate_group_errors(test_analysis_df, 'is_old_building', 'Old Building: '))

error_by_category_df = pd.DataFrame(category_summaries)
print('\n=== Error Analysis by Category ===')
display(error_by_category_df)
```

```python
# ------------------------------------------------------------------------
# QUESTION 3: SALES-SPEED CLASSIFICATION
# ------------------------------------------------------------------------

# Target for classification
y_cls = df['sold_within_45_days'].copy()
# Features are the same as regression
X_cls_raw = df[predictors_reg].copy()

# Clean amenity score
if X_cls_raw['amenity_score'].isnull().sum() > 0:
    X_cls_raw['amenity_score'] = X_cls_raw['amenity_score'].fillna(X_cls_raw['amenity_score'].median())

# One-hot encoding for classification features
X_cls = pd.get_dummies(X_cls_raw, columns=categorical_cols, drop_first=True, dtype=float)

# Train test split
X_train_cls, X_test_cls, y_train_cls, y_test_cls = train_test_split(
    X_cls, y_cls, test_size=0.20, random_state=random_state
)

# Initialize and train Decision Tree Classifier
# Let's use max_depth to avoid extreme overfitting and ensure business interpretability
dt_model = DecisionTreeClassifier(max_depth=5, random_state=random_state)
dt_model.fit(X_train_cls, y_train_cls)

# Dummy baseline classifier
dummy_model = DummyClassifier(strategy='most_frequent')
dummy_model.fit(X_train_cls, y_train_cls)

# Predictions
dt_preds = dt_model.predict(X_test_cls)
dummy_preds = dummy_model.predict(X_test_cls)

# Evaluation metrics
def get_classification_metrics(y_true, y_pred, label_pos=1):
    acc = accuracy_score(y_true, y_pred)
    # We treat sold_within_45_days = 0 (not sold / slow) as the slow class, or
    # are we predicting slow or fast? The question asks about identifying listings likely to sell slowly.
    # So, we can look at metrics for class 0 (slow-moving) as the target of interest, but standard metrics
    # are often computed for class 1 or both. Let us print metrics for both classes and clearly display them.
    # Let's compute specifically for 'slow-moving / at-risk' (class 0).
    prec_0 = precision_score(y_true, y_pred, pos_label=0)
    rec_0 = recall_score(y_true, y_pred, pos_label=0)
    f1_0 = f1_score(y_true, y_pred, pos_label=0)

    # Also regular metrics (pos_label=1)
    prec_1 = precision_score(y_true, y_pred, pos_label=1, zero_division=0)
    rec_1 = recall_score(y_true, y_pred, pos_label=1, zero_division=0)
    f1_1 = f1_score(y_true, y_pred, pos_label=1, zero_division=0)

    return acc, prec_0, rec_0, f1_0, prec_1, rec_1, f1_1

dt_metrics = get_classification_metrics(y_test_cls, dt_preds)
dummy_metrics = get_classification_metrics(y_test_cls, dummy_preds)

print('=== Classification Performance Table ===')
print(f'Test Class 0 (Slow-moving) Distribution in Test Set: {y_test_cls.value_counts()}')

metrics_summary_df = pd.DataFrame({
    'Metric': ['Accuracy', 'Precision (Slow)', 'Recall (Slow)', 'F1-Score (Slow)', 'Precision (Fast)', 'Recall (Fast)', 'F1-Score (Fast)'],
    'Majority Baseline': [
        dummy_metrics[0], dummy_metrics[1], dummy_metrics[2], dummy_metrics[3], dummy_metrics[4], dummy_metrics[5], dummy_metrics[6]
    ],
    'Decision Tree': [
        dt_metrics[0], dt_metrics[1], dt_metrics[2], dt_metrics[3], dt_metrics[4], dt_metrics[5], dt_metrics[6]
    ]
})
display(metrics_summary_df)

print('\n=== Confusion Matrices ===')
print('Decision Tree Confusion Matrix:')
print(confusion_matrix(y_test_cls, dt_preds))
print('Majority Baseline Confusion Matrix:')
print(confusion_matrix(y_test_cls, dummy_preds))
```

```python
# ------------------------------------------------------------------------
# QUESTION 4: RANDOM FOREST ALTERNATIVE
# ------------------------------------------------------------------------

rf_model = RandomForestClassifier(n_estimators=100, random_state=random_state)
rf_model.fit(X_train_cls, y_train_cls)
rf_preds = rf_model.predict(X_test_cls)

rf_metrics = get_classification_metrics(y_test_cls, rf_preds)

comparison_df = pd.DataFrame({
    'Metric': ['Accuracy', 'Precision (Slow)', 'Recall (Slow)', 'F1-Score (Slow)', 'Precision (Fast)', 'Recall (Fast)', 'F1-Score (Fast)'],
    'Decision Tree': [
        dt_metrics[0], dt_metrics[1], dt_metrics[2], dt_metrics[3], dt_metrics[4], dt_metrics[5], dt_metrics[6]
    ],
    'Random Forest': [
        rf_metrics[0], rf_metrics[1], rf_metrics[2], rf_metrics[3], rf_metrics[4], rf_metrics[5], rf_metrics[6]
    ]
})
print('=== Decision Tree vs Random Forest ===')
display(comparison_df)

print('\n=== Confusion Matrix Comparison ===')
print('Decision Tree:')
print(confusion_matrix(y_test_cls, dt_preds))
print('Random Forest:')
print(confusion_matrix(y_test_cls, rf_preds))
```
